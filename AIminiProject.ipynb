```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract the coefficients in the form f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Define the prediction function
def f(x_value):
    return mymodel(x_value)

# Print the prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values for prediction
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    "time": [
        "08:00", "09:00", "10:00", "11:00",
        "12:00", "13:00", "14:00", "15:00",
        "16:00", "18:00", "19:00", "20:00"
    ],
    "speed": [
        52, 49, 55, 58,
        60, 57, 54, 50,
        46, 43, 40, 45
    ]
}

df = pd.DataFrame(data)

# Convert time into hour
df["hour"] = pd.to_datetime(df["time"]).dt.hour

# Independent variable
X = df[["hour"]]

# Dependent variable
y = df["speed"]

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Predict speed at 17:00
predicted_speed = model.predict([[17]])[0]

# 1. Print predicted speed
print(f"Predicted speed at 17:00: {predicted_speed:.2f} km/h")

# Predictions for the existing data
df["predicted_speed"] = model.predict(X)

# 2. Display prediction on graph
plt.figure(figsize=(10, 6))

plt.scatter(
    df["hour"],
    df["speed"],
    label="Observed speed"
)

plt.plot(
    df["hour"],
    df["predicted_speed"],
    label="Regression line"
)

# 17:00 prediction
plt.scatter(
    17,
    predicted_speed,
    s=100,
    label=f"17:00 prediction = {predicted_speed:.2f} km/h"
)

plt.annotate(
    f"{predicted_speed:.2f} km/h",
    (17, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.xlabel("Time")
plt.ylabel("Speed (km/h)")
plt.title("Vehicle Speed Prediction at 17:00")
plt.xticks(range(8, 21))
plt.legend()
plt.grid(True)

plt.show()
```
